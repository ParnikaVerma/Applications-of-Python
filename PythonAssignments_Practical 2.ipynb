```python
# 1. Create an empty list
shopping_list = []
print(shopping_list)
```

```python
# 2. Add items to the shopping list
shopping_list = ["Watch", "Laptop", "Shoes", "Pen", "Clothes"]
```

```python
print(shopping_list)
```

```python
# 3. Add Football to the shopping list
shopping_list.append("Football")
print(shopping_list)
```

```python
# 4. Print the last item
print("Last item:", shopping_list[-1])
```

```python
# 5. Print the entire shopping list
print("Shopping list:", shopping_list)
```

```python
# 6. Print important items: Laptop and Shoes
print("Important items:", shopping_list[1], shopping_list[2])
```

```python
# 7. Change Pen to Notebook
shopping_list[3] = "Notebook"
print(shopping_list)
```

```python
# 8. Delete the least important item (Clothes)
shopping_list.remove("Clothes")
```

```python
print("Final shopping list:", shopping_list)
```
