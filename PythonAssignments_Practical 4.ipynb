```python
players = [
    {"name": "Serena Williams", "sport": "Tennis", "count": 23, "award": "Grand Slams"},
    {"name": "Lionel Messi", "sport": "Soccer", "count": 7, "award": "Ballon d'Ors"},
    {"name": "Michael Phelps", "sport": "Swimming", "count": 23, "award": "Gold Medals"},
    {"name": "Usain Bolt", "sport": "Athletics", "count": 8, "award": "Gold Medals"},
    {"name": "Roger Federer", "sport": "Tennis", "count": 20, "award": "Grand Slams"},
    {"name": "Cristiano Ronaldo", "sport": "Soccer", "count": 5, "award": "Ballon d'Ors"},
]

#QUES !
for p in players:
    if p["name"] == "Lionel Messi":
        if p["count"] > 10:
            print(p["name"], "-", p["sport"], "-", p["count"], p["award"])
        else:
            print(p["name"], "does not have more than 10 achievements")
```

```python
#QUES 4
num = int(input("Enter a number between 1 and 9999: "))

if 1 <= num <= 9999:
    # even or odd
    if num % 2 == 0:
        print(num, "is even")
    else:
        print(num, "is odd")

    # palindrome
    if str(num) == str(num)[::-1]:
        print(num, "is a palindrome")
    else:
        print(num, "is not a palindrome")

    # divisible by sum of digits
    digit_sum = 0
    for d in str(num):
        digit_sum = digit_sum + int(d)

    if num % digit_sum == 0:
        print(num, "is divisible by the sum of its digits", digit_sum)
    else:
        print(num, "is not divisible by the sum of its digits", digit_sum)
else:
    print("Number is out of range")
```

```python
squares = ['orange', 'orange', 'purple', 'blue', 'orange']
new_squares = []

i = 0
while i < len(squares) and squares[i] == 'orange':
    new_squares.append(squares[i])
    i = i + 1

print(new_squares)
```

```python
Animals = ["lion", "giraffe", "gorilla", "parrots", "crocodile", "deer", "swan"]
seven_letter = []

i = 0
while i < len(Animals):
    if len(Animals[i]) == 7:
        seven_letter.append(Animals[i])
    i = i + 1

print(seven_letter)
```
