```python
# Problem 4- creating the dataset (10 records, 6 columns) with two missing values
data = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Aman", "Priya", "Rohit", "Sneha", "Karan", "Neha", "Vikram", "Anjali", "Arjun", "Meera"],
    "Age": [25, 34, 28, np.nan, 41, 36, 23, 30, 45, 29],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female"],
    "City": ["Delhi", "Mumbai", "Shimla", "Chandigarh", "Pune", "Jaipur", "Delhi", "Mumbai", "Pune", "Shimla"],
    "Purchase_Amount": [2500, 1800, np.nan, 3200, 1500, 2700, 4100, 2200, 3600, 1900]
})

# first 5 rows
data.head()
```

```python
# summary statistics
data.describe()
```

```python
# checking missing values
data.isnull().sum()
```

```python
# checking missing values
data.isnull().sum()
```

```python
# handling missing values
# Age -> filled with median, Purchase_Amount -> filled with mean
data["Age"] = data["Age"].fillna(data["Age"].median()).astype(int)
data["Purchase_Amount"] = data["Purchase_Amount"].fillna(data["Purchase_Amount"].mean()).round(2)

# checking again
print(data.isnull().sum())
data
```

```python
# Problem 5- customers with Age greater than 30
data[data["Age"] > 30]
```

```python
# total purchase amount for male and female separately
data.groupby("Gender")["Purchase_Amount"].sum()
```

```python
# selecting only 3 columns
data[["Customer_ID", "Name", "Purchase_Amount"]]
```

```python
# PROBLEM 6- adding Discounted_Price column (10% discount)
data["Discounted_Price"] = data["Purchase_Amount"] * 0.90
data
```

```python
# average purchase amount by gender
data.groupby("Gender")["Purchase_Amount"].mean()
```

```python
# sorting by PurchaseAmount in descending order
data.sort_values(by="Purchase_Amount", ascending=False)
```
