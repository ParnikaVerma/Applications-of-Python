```python
#PROBLEM 1- ZIPPER FILE 
def zipper_file(filename, strings):
    file = open(filename, "w")

    for i in range(0, len(strings), 2):
        s1 = strings[i]
        s2 = strings[i + 1]

        zipped = ""

        for j in range(len(s1)):
            zipped += s1[j] + s2[j]

        file.write(zipped + "\n")

    file.close()

#Implementation
words = ["abc", "xyz", "mno", "pqr"]
zipper_file("output.txt", words)

print("File created successfully.")

```

```python
#PROBLEM 2- REFORMAT DATE 
def reformat_date(date_list):
    months = {
        "01": "January",
        "02": "February",
        "03": "March",
        "04": "April",
        "05": "May",
        "06": "June",
        "07": "July",
        "08": "August",
        "09": "September",
        "10": "October",
        "11": "November",
        "12": "December"
    }
    for date in date_list:
        parts = date.split("/")
        month = months[parts[0]]
        day = parts[1]
        year = parts[2]
        print("Office Hours on", month, day + ",", year)


# Implementation:
dates = [
    "01/15/2024",
    "05/08/2025",
    "12/25/2026"
]

reformat_date(dates)
```
