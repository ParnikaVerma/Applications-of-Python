```python
#IMPORTING LIBRARIES 

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Iris Dataset
iris = sns.load_dataset("iris")

print(iris.head())
```

```python
#SCATTER PLOT 

sns.scatterplot(
    data=iris,
    x="sepal_length",
    y="sepal_width",
    hue="species"
)

plt.title("Scatter Plot of Sepal Length vs Sepal Width")
plt.show()

#ANALYSIS
#The scatter plot shows the relationship between sepal length and sepal width.
#Setosa samples form a separate cluster from the other species.
#Versicolor and Virginica overlap slightly.
#Sepal length generally increases with species variation.
```

```python
#DENSITY PLOT

sns.kdeplot(
    data=iris,
    x="petal_length",
    hue="species",
    fill=True
)

plt.title("Density Plot of Petal Length")
plt.show()

#Analysis
#The density plot shows the distribution of petal length.
#Setosa has the smallest petal lengths.
#Virginica has the largest petal lengths.
#The distributions are clearly separated, making petal length useful for classification.
```

```python
# VIOLIN PLOT 

sns.violinplot(
    data=iris,
    x="species",
    y="petal_width"
)

plt.title("Violin Plot of Petal Width")
plt.show()

#Analysis 
#The violin plot displays both distribution and density of petal width.
#Setosa has very small petal widths with little variation.
#Virginica shows larger petal widths and greater spread.
#The shape of each violin indicates how the values are distributed.
```

```python
#HEATMAP 

corr = iris.select_dtypes(include="number").corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Heatmap")
plt.show()

#Analysis 
#Petal length and petal width have a very strong positive correlation.
#Sepal length is positively correlated with petal length and petal width.
#Sepal width has a weaker correlation with the other features.
#Strong positive correlations indicate that the variables tend to increase together.
#Petal measurements are the most important features for distinguishing iris species.
```

```python
#JOINT PLOT 

sns.jointplot(
    data=iris,
    x="petal_length",
    y="petal_width",
    kind="scatter"
)
plt.show()

#Analysis 
#The joint plot combines scatter and distribution plots.
#Petal length and petal width show a strong positive relationship.
#As petal length increases, petal width also increases.
#Most observations are concentrated around the center of the distribution.
```

```python
#PAIR PLOT 

sns.pairplot(
    iris,
    hue="species"
)
plt.show()

#Analysis 
#The pair plot compares all numerical features simultaneously.
#Setosa is clearly separated from the other species in most feature combinations.
#Petal length and petal width provide the best separation between species.
#Some overlap exists between Versicolor and Virginica.
#Pair plots help identify patterns, relationships, and clustering in the dataset.
```
