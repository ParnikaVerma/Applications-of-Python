```python
# Create a tuple of product prices
prices = (250, 300, 150, 400, 100, 350, 200)
```

```python
# Find the highest and lowest price
print("Highest price:", max(prices))
print("Lowest price:", min(prices))
```

```python
# Find the total of all prices
print("Total price:", sum(prices))
```

```python
# Convert tuple into sorted list
sorted_prices = sorted(prices)
print("Prices in ascending order:", sorted_prices)
```

```python
# Try to change an element
prices[0] = 500

#This line will give an error, SINCE A tuple cannot be changed after it is created. So when we try to change prices[0], Python gives a TypeError
#This is because tuples are immutable, whereas lists can be modified
```
